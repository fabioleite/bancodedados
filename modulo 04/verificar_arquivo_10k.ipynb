# Verificação e geração do arquivo de 10 mil registros

Este notebook valida se o arquivo de 10 mil registros já existe e, caso não exista, gera e salva o dataset em disco no diretório do projeto.

```python
from pathlib import Path
import csv
from datetime import datetime

workspace = Path.cwd()
output_dir = workspace / 'modulo 04' / 'dados'
output_dir.mkdir(parents=Tpython --version
rue, exist_ok=True)
arquivo = output_dir / 'divergencia_nfe_efd_202401.csv'
print(f'Diretório do projeto: {workspace}')
print(f'Arquivo de destino: {arquivo}')
print(f'Existe antes da geração? {arquivo.exists()}')
```

```python
campos = [
    'periodo_apuracao',
    'cnpj',
    'uf_origem',
    'uf_destino',
    'cfop',
    'valor_documento',
    'valor_icms',
    'tipo_operacao',
    'numero_doc',
    'chave_acesso',
    'situacao_documento',
    'fonte_dado',
]

def gerar_registro(i: int):
    cnpjs = [
        '12345678000199', '98765432000155', '11122233000144', '22233344000166',
        '33344455000177', '44455566000188', '55566677000199', '66677788000110',
        '77788899000121', '88899900000132'
    ]
    ufs = ['SP', 'RJ', 'MG', 'PR', 'SC', 'RS', 'GO', 'ES', 'MT', 'MS']
    cfops = ['6102', '6108', '5405', '5403', '5102', '6203', '6205', '5305']
    cnpj = cnpjs[(i - 1) % len(cnpjs)]
    uf_origem = ufs[(i - 1) % len(ufs)]
    cfop = cfops[(i - 1) % len(cfops)]
    valor_documento = round(1500 + ((i * 8.75) % 30000), 2)
    valor_icms = round(valor_documento * 0.18, 2)
    tipo_operacao = 'SAIDA' if i % 2 == 0 else 'ENTRADA'
    numero_doc = 50000 + i
    chave = f"352401{cnpj}{(i % 10000):05d}000000000{(i % 1000):03d}00000000"[:44]
    situacao = 'CANCELADA' if i % 5 == 0 else 'AUTORIZADA'
    return [
        '202401', cnpj, uf_origem, 'PB', cfop,
        f'{valor_documento:.2f}', f'{valor_icms:.2f}',
        tipo_operacao, numero_doc, chave, situacao, 'SISTEMA_EXTERNO'
    ]

print('Schema definido:')
print(campos)
print('\nExemplo de registro gerado:')
print(gerar_registro(1))
```

```python
if arquivo.exists():
    print(f'O arquivo já existe em: {arquivo}')
    with arquivo.open('r', encoding='utf-8', newline='') as f:
        reader = csv.reader(f, delimiter=';')
        rows = list(reader)
    print(f'Linhas no arquivo: {len(rows) - 1}')
    print('Primeiras 3 linhas:')
    for row in rows[:4]:
        print(row)
else:
    with arquivo.open('w', newline='', encoding='utf-8') as f:
        writer = csv.writer(f, delimiter=';')
        writer.writerow(campos)
        for i in range(1, 10001):
            writer.writerow(gerar_registro(i))
    print(f'Arquivo gerado em: {arquivo}')
    with arquivo.open('r', encoding='utf-8', newline='') as f:
        reader = csv.reader(f, delimiter=';')
        rows = list(reader)
    print(f'Linhas no arquivo após geração: {len(rows) - 1}')
    print('Primeiras 3 linhas do arquivo gerado:')
    for row in rows[:4]:
        print(row)
```

```python
with arquivo.open('r', encoding='utf-8', newline='') as f:
    reader = csv.reader(f, delimiter=';')
    rows = list(reader)
    count = len(rows) - 1
    print(f'Quantidade confirmada: {count} registros')
    print('Amostra final:')
    for row in rows[-3:]:
        print(row)
print('Validação concluída com sucesso.')
```
